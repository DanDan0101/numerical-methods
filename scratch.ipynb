```python
import numpy as np
from matplotlib import pyplot as plt
from numba import njit
from tqdm.notebook import tqdm

from scipy.integrate import solve_ivp
```

```python
abs(np.array([1, -1]))
```

```python
def harm(_, y):
    return np.array([y[1], -y[0]])
```

```python
rk_sol = solve_ivp(harm, (0, 1000), np.array([0, 1]), method = "RK45")
t = rk_sol.t
true_sol = np.sin(t)
```

```python
plt.subplots(dpi = 150, layout = "constrained")
plt.plot(t, np.abs(rk_sol.y[0] - true_sol), label = "RK5(4)")
plt.legend()
plt.ylabel("$f(t)$")
plt.xlabel("$t$")
plt.title("Absolute Error")
```

```python
plt.subplots(dpi = 150, layout = "constrained")
plt.plot(t, rk_sol.y[0], label = "RK5(4)")
plt.plot(t, true_sol, label = "True")
plt.legend()
plt.ylabel("$f(t)$")
plt.xlabel("$t$")
plt.title("Simple Harmonic Oscillator")
```
